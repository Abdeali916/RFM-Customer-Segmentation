```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_excel(r"Online Retail.xlsx")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
df.shape
```

```python
df[df['InvoiceNo'].astype(str).str.startswith('C')].head()
```

```python
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]
df
```

```python
df = df.dropna(subset=['CustomerID'])
df
```

```python
df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)]
```

```python
df['Revenue'] = df['Quantity'] * df['UnitPrice']
```

```python
df[['Quantity', 'UnitPrice', 'Revenue']].head()
```

```python
df['CustomerID'] = df['CustomerID'].astype(int)
```

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nNegative quantities:", (df['Quantity'] < 0).sum())

print("\nInvalid prices:", (df['UnitPrice'] <= 0).sum())
```

# Exploratory Data Analysis (EDA)

```python
print("Total Revenue:", df['Revenue'].sum())
print("Total Orders:", df['InvoiceNo'].nunique())
print("Total Customers:", df['CustomerID'].nunique())
print("Total Products:", df['StockCode'].nunique())
print("Total Countries:", df['Country'].nunique())
```

```python
print("Start Date:", df['InvoiceDate'].min())
print("End Date:", df['InvoiceDate'].max())
```

```python
df['Revenue'].describe()
```

```python
print("Average Order Value:", 
      df.groupby('InvoiceNo')['Revenue'].sum().mean())
```

```python
df['Month'] = df['InvoiceDate'].dt.to_period('M')
df
```

```python
monthly_revenue = (
    df.groupby('Month')['Revenue']
      .sum()
)

monthly_revenue
```

```python
monthly_revenue.plot(
    kind='line',
    figsize=(12,5),
    title='Monthly Revenue'
)
```

```python
monthly_orders = (
    df.groupby('Month')['InvoiceNo']
      .nunique()
)

monthly_orders
```

```python
monthly_orders.plot(
    kind='line',
    figsize=(12,5),
    title='Monthly Orders'
)
```

```python
top_products = (
    df.groupby('Description')['Revenue']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_products
```

```python
top_products.sort_values().plot(
    kind='barh',
    figsize=(10,6),
    title='Top 10 Products by Revenue'
)
```

```python
top_quantity_products = (
    df.groupby('Description')['Quantity']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_quantity_products
```

```python
country_revenue = (
    df.groupby('Country')['Revenue']
      .sum()
      .sort_values(ascending=False)
)

country_revenue.head(10)
```

```python
country_revenue.head(10).sort_values().plot(
    kind='barh',
    figsize=(10,6),
    title='Top 10 Countries by Revenue'
)
```

```python
customer_revenue = (
    df.groupby('CustomerID')['Revenue']
      .sum()
      .sort_values(ascending=False)
)

customer_revenue.head(10)
```

```python
customer_revenue.head(10)
```

```python
customer_orders = (
    df.groupby('CustomerID')['InvoiceNo']
      .nunique()
      .sort_values(ascending=False)
)

customer_orders.head(10)
```

```python
customer_orders.describe()
```

```python
customer_orders.plot(
    kind='hist',
    bins=30,
    figsize=(10,5),
    title='Distribution of Orders per Customer'
)
```
